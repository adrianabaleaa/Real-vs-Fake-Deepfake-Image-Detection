```python
import tensorflow as tf

# 1. LOAD MODEL
cale_model_salvat = '/kaggle/input/notebooks/adriananicoletabalea/deepfake/deepfake_detector.keras' 
model = tf.keras.models.load_model(cale_model_salvat)


# 2. PATHS
cale_test = '/kaggle/input/datasets/xhlulu/140k-real-and-fake-faces/real_vs_fake/real-vs-fake/test'


test_dataset = tf.keras.utils.image_dataset_from_directory(
    cale_test,
    image_size=(224, 224), 
    batch_size=32,         
    label_mode='binary',   
    shuffle=False          
)


AUTOTUNE = tf.data.AUTOTUNE
test_dataset = test_dataset.prefetch(buffer_size=AUTOTUNE)

# 3. Test
rezultate = model.evaluate(test_dataset)

# 4. Show results
loss_test = rezultate[0]
acuratete_test = rezultate[1]

print("\n" + "="*45)
print("       RESULTS (TEST SET)")
print("="*45)
print(f"ACCURACY : {acuratete_test * 100:.2f}%")
print(f"LOSS        : {loss_test:.4f}")
print("="*45)
```
